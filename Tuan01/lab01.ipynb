### Bài 1. Tạo Tensor cơ bản

```python
import torch

a = torch.tensor(
    [[1, 2, 3],
     [4, 5, 6]], dtype=torch.int32
)
print("Tensor a:\n", a)

# YÊU CẦU: Cho biết kích thước, số chiều và kiểu dữ liệu
print("\n--- Yêu cầu Bài 1 ---")
print("Kích thước (Shape):", a.shape)
print("Số chiều (ndim):", a.ndim)
print("Kiểu dữ liệu (dtype):", a.dtype)
```

### Bài 2. Tạo các Tensor thường dùng

```python
import torch

a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)

print("torch.rand:\n", a)
print("torch.randn:\n", b)
print("torch.zeros:\n", c)
print("torch.ones:\n", d)
print("torch.eye:\n", e)

# YÊU CẦU: Tạo thêm Tensor 5x5 chứa toàn giá trị 10
f = torch.full((5, 5), 10)
print("\nTensor 5x5 toàn giá trị 10:\n", f)
```

- `torch.randn()`: Sinh ngẫu nhiên các giá trị từ phân phối chuẩn tắc (standard normal distribution) với trung bình = 0 và độ lệch chuẩn = 1.

### Bài 4. Slicing và Indexing

```python
import torch

a = torch.randn(3, 4, 5)
print("Original tensor shape:", a.shape)

print("\nTensor a[1] shape:", a[1].shape)
print("Tensor a[1:, 2:4] shape:", a[1:, 2:4].shape)

# YÊU CẦU: Thử lấy hai phần tử cuối cùng trên chiều thứ ba
# Lấy tất cả các phần tử ở chiều 1 và chiều 2, nhưng chỉ lấy 2 cột cuối ở chiều 3
last_two = a[:, :, -2:]
print("\nHai phần tử cuối cùng trên chiều thứ ba (shape):", last_two.shape)
print(last_two)
```

### Bài 6. Phép toán và nhân ma trận

```python
import torch

a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:\n", a)
print("b:\n", b)
print("\na + b:\n", a + b)
print("\na ** 2:\n", a ** 2)
print("\na / b:\n", a / b)

print("\nMatrix multiplication (a @ b.T):\n", a @ b.T)
print("Shape của kết quả a @ b.T:", (a @ b.T).shape)
```

**Nhận xét (Yêu cầu Bài 6):**
- Phép toán `a * b`: Là phép nhân từng phần tử tương ứng (element-wise multiplication) của hai ma trận cùng kích thước.
- Phép toán `a @ b.T`: Là phép nhân ma trận (matrix multiplication). Kích thước của `a` là (2, 3), `b.T` là (3, 2) nên kết quả có shape là (2, 2).

### Bài 7. Các hàm thống kê trên Tensor

```python
import torch

a = torch.randn(3, 4)
print("Original:\n", a)

print("\nMean (dim=0):")
print(torch.mean(a, dim=0))

print("\nStandard deviation (dim=0):")
print(torch.std(a, dim=0))

print("\nCumulative sum (dim=0):")
print(torch.cumsum(a, dim=0))

# YÊU CẦU: Thay dim=0 bằng dim=1
print("\n--- Yêu cầu Bài 7 ---")
print("Mean (dim=1):\n", torch.mean(a, dim=1))
print("Standard deviation (dim=1):\n", torch.std(a, dim=1))
print("Cumulative sum (dim=1):\n", torch.cumsum(a, dim=1))
```

**Nhận xét (Yêu cầu Bài 7):**
- Khi dùng `dim=0`, các hàm thống kê sẽ tính toán theo chiều dọc (gom các hàng lại), tức là tính theo từng cột. Kết quả trả về gồm 4 giá trị.
- Khi dùng `dim=1`, các hàm thống kê sẽ tính toán theo chiều ngang (gom các cột lại), tức là tính trên từng hàng riêng biệt. Kết quả trả về gồm 3 giá trị.

### Bài 8. Autograd cơ bản

```python
import torch

# Phần 1: y = x^2
x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()

print("--- Phần y = x^2 ---")
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

# Phần 2: YÊU CẦU: Đổi hàm thành y = 3x^2 + 2x + 1
x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * (x2 ** 2) + 2 * x2 + 1
y2.backward()

print("\n--- Phần y = 3x^2 + 2x + 1 ---")
print("x2 =", x2)
print("y2 =", y2)
print("x2.grad =", x2.grad)
```

**Tính thủ công (Yêu cầu Bài 8):**
1. Hàm $y = x^2$ -> Đạo hàm $y' = 2x$. Tại $x = 3.6$ -> $y' = 2 \times 3.6 = 7.2$. Kết quả khớp với `x.grad`.
2. Hàm $y = 3x^2 + 2x + 1$ -> Đạo hàm $y' = 6x + 2$. Tại $x = 3.6$ -> $y' = 6 \times 3.6 + 2 = 23.6$. Kết quả khớp với `x2.grad`.

### Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch

# Standard polynomial y = x^2 + 2x + 3
polynomial = np.poly1d([1, 2, 3])

N = 20
X = np.random.randn(N, 1)
Y = polynomial(X)

XX = np.hstack((X * X, X, np.ones_like(X)))

x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

w = torch.randn(3, 1, requires_grad=True)
print("Initial coefficients:\n", w)

optimizer = torch.optim.NAdam([w], lr=0.01)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y_pred - y))
    mse.backward()
    optimizer.step()

print("\nLearned coefficients:\n", w)
```

**Nhận xét (Yêu cầu Bài 9):**
- Ba hệ số học được rất xấp xỉ với hệ số gốc `[1, 2, 3]` của đa thức (sẽ có sai số cực nhỏ do bản chất xấp xỉ của Gradient Descent).
- **Vai trò của các hàm optimizer:**
  - `optimizer.zero_grad()`: Xóa sạch (reset về 0) các gradient đã được tích lũy từ vòng lặp trước. Nếu không, gradient sẽ bị cộng dồn sai lệch.
  - `mse.backward()`: Kích hoạt cơ chế Autograd tính đạo hàm ngược, tìm ra gradient của hàm loss (`mse`) theo từng trọng số (`w`).
  - `optimizer.step()`: Cập nhật giá trị mới cho các trọng số (`w`) dựa vào gradient vừa tính được và tốc độ học (`lr`) để giảm dần sai số.

### Bài 10. Giải hệ phương trình bằng Autograd

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    optimizer.zero_grad()
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    sqerr.backward()
    optimizer.step()

print(f"Nghiệm tìm được: A = {A.item():.4f}, B = {B.item():.4f}, C = {C.item():.4f}, D = {D.item():.4f}")

print("\n--- Kiểm tra lại phương trình ---")
print(f"A + B = {A.item() + B.item():.4f} (Đích: 9)")
print(f"C - D = {C.item() - D.item():.4f} (Đích: 1)")
print(f"A + C = {A.item() + C.item():.4f} (Đích: 8)")
print(f"B - D = {B.item() - D.item():.4f} (Đích: 2)")
```

**Giải thích (Yêu cầu Bài 10):**
Bài toán này có **vô số nghiệm** (nhiều nghiệm) bởi vì các phương trình trong hệ phụ thuộc tuyến tính lẫn nhau. 
Lấy phương trình (1) + (2): $(A + B) + (C - D) = 9 + 1 = 10 \Rightarrow A + B + C - D = 10$
Lấy phương trình (3) + (4): $(A + C) + (B - D) = 8 + 2 = 10 \Rightarrow A + B + C - D = 10$
Thực chất chỉ có 3 phương trình độc lập nhưng lại có 4 ẩn, dẫn tới hệ có một bậc tự do (đường thẳng các nghiệm). Việc Gradient Descent dừng lại ở nghiệm nào phụ thuộc hoàn toàn vào giá trị khởi tạo ngẫu nhiên ban đầu của A, B, C, D.

### Bài 11. Các yêu cầu tổng hợp

```python
import torch
import numpy as np

print("--- 11.1 Tạo Tensor X và Reshape ---")
X = torch.arange(24) # Tạo tensor có 24 phần tử
print("Reshape 4x6:\n", X.reshape(4, 6))
print("Reshape 2x12:\n", X.reshape(2, 12))
print("Reshape 2x3x4:\n", X.reshape(2, 3, 4))

print("\n--- 11.2 Các phép toán trên A, B (3x4) ---")
A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("A + B:\n", A + B)
print("A * B:\n", A * B)
print("A / B:\n", A / B)
print("Trung bình theo cột (A):\n", torch.mean(A, dim=0))
print("Độ lệch chuẩn theo cột (A):\n", torch.std(A, dim=0))

print("\n--- 11.3 Autograd cho y = 2x^2 + 5x + 3 ---")
x = torch.tensor(2.0, requires_grad=True)
y = 2 * (x**2) + 5*x + 3
y.backward()
print("PyTorch dy/dx tại x=2:", x.grad.item())
# Tính thủ công: y' = 4x + 5. Tại x = 2 -> 4(2)+5 = 13
print("Thủ công dy/dx tại x=2: 13") 

print("\n--- 11.4 Huấn luyện với y = 2x^2 - 3x + 5 ---")
poly_new = np.poly1d([2, -3, 5])
Y_new = poly_new(X_np := np.random.randn(20, 1))
XX_new = np.hstack((X_np * X_np, X_np, np.ones_like(X_np)))

x_t = torch.tensor(XX_new, dtype=torch.float32)
y_t = torch.tensor(Y_new, dtype=torch.float32)
w_t = torch.randn(3, 1, requires_grad=True)

opt = torch.optim.NAdam([w_t], lr=0.01)
for _ in range(1500):
    opt.zero_grad()
    mse = torch.mean(torch.square(x_t @ w_t - y_t))
    mse.backward()
    opt.step()

print("Hệ số tìm được (kỳ vọng [2, -3, 5]):\n", w_t.detach().numpy())
```
